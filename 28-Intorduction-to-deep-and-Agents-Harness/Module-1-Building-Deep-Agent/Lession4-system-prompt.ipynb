## **The System Prompt**
- The system prompt guides the model and is part of every model call.
- It is re-sent on every turn, ahead of the conversation

## **Prompt Caching**
<p>Prompt caching lets a model provider reuse the work it already did on a prompt instead of reprocessing the same tokens on every request. Agent loops are the ideal case for this: the system prompt and tool definitions rarely change between turns, so a large chunk of every request is identical to the request before it.</p>

Caching turns that repetition into a steep discount instead of a full-price repeat. As Manus AI put it, "the KV-cache hit rate is the single most important metric for a production-stage AI agent."
