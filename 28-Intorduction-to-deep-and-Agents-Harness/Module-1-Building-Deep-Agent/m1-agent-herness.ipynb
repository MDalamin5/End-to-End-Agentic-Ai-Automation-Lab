## What is `Harness`
- Harness is everything the model connect the real world.
- Everything around the model help to complete task
    - skill
    - memory
    - system prompt
    - subagents
    - context
    - tools
## What is the job of a harness
- To get the model the right context at the right time for the given task.

## Why do you need a harness
- Work in the enviroment where the can take actions
- Connect to your data
- Mange growing context over long run
- HITL
- Connect to our data
- Parallelize tas
- Improve over time


## What is deep agent
- A customizable agent harness purpose build for complex, real-world tasks.
- **Backbone:** Execution Environment
    - Code interpreter
    - Sandbox (skill, memory, summarization, context offloading, prompt caching)
    - Filesystem
- **Backbone:** Context Management
    - skill, memory, summarization, context offloading, prompt caching
- **Backbone:** Delegation
